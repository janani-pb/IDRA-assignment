```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.model_selection import train_test_split
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.impute import SimpleImputer
from sklearn.linear_model import LinearRegression, LogisticRegression
from sklearn.metrics import (
    mean_absolute_error, mean_squared_error, r2_score,
    accuracy_score, precision_score, recall_score, f1_score,
    confusion_matrix, classification_report
)

pd.set_option("display.max_columns", None)
pd.set_option("display.float_format", lambda x: f"{x:.3f}")

file_path = "/mnt/data/Day18_19_student_habits_performance.csv"
df = pd.read_csv("Day18_19_student_habits_performance.csv")

print("=" * 80)
print("STUDENT PERFORMANCE - ML PREDICTION AND MODEL EVALUATION")
print("=" * 80)

print("\n1. DATASET OVERVIEW")
print("-" * 80)
print("Dataset Shape:", df.shape)
print("\nFirst 5 Rows:")
display(df.head())
print("\nDataset Information:")
df.info()

print("\n2. DESCRIPTIVE STATISTICS")
print("-" * 80)
display(df.describe(include="all").T)

print("\n3. NUMERICAL AND CATEGORICAL VARIABLES")
print("-" * 80)
numerical_columns = df.select_dtypes(include=np.number).columns.tolist()
categorical_columns = df.select_dtypes(include="object").columns.tolist()

print("Numerical Variables:")
print(numerical_columns)

print("\nCategorical Variables:")
print(categorical_columns)

print("\n4. MISSING VALUE ANALYSIS")
print("-" * 80)
missing_values = df.isnull().sum()
missing_percentage = (df.isnull().sum() / len(df)) * 100
missing_table = pd.DataFrame({
    "Missing Values": missing_values,
    "Percentage": missing_percentage
})
display(missing_table[missing_table["Missing Values"] > 0])

print("\nDuplicate Rows:", df.duplicated().sum())

print("\n5. UNIQUE VALUES IN CATEGORICAL VARIABLES")
print("-" * 80)
for col in categorical_columns:
    print(f"\n{col}:")
    print(df[col].value_counts(dropna=False))

print("\n6. DISTRIBUTION OF NUMERICAL VARIABLES")
print("-" * 80)

for col in numerical_columns:
    plt.figure(figsize=(8, 4))
    sns.histplot(df[col], kde=True)
    plt.title(f"Distribution of {col}")
    plt.xlabel(col)
    plt.ylabel("Frequency")
    plt.show()

print("\n7. BOXPLOTS FOR OUTLIER DETECTION")
print("-" * 80)

for col in numerical_columns:
    plt.figure(figsize=(8, 3))
    sns.boxplot(x=df[col])
    plt.title(f"Boxplot of {col}")
    plt.xlabel(col)
    plt.show()

print("\n8. OUTLIER ANALYSIS USING IQR")
print("-" * 80)

outlier_results = []

for col in numerical_columns:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    outliers = df[(df[col] < lower_bound) | (df[col] > upper_bound)]

    outlier_results.append([
        col,
        Q1,
        Q3,
        IQR,
        lower_bound,
        upper_bound,
        len(outliers)
    ])

outlier_table = pd.DataFrame(
    outlier_results,
    columns=[
        "Variable", "Q1", "Q3", "IQR",
        "Lower Bound", "Upper Bound", "Outlier Count"
    ]
)

display(outlier_table)

print("\n9. CORRELATION ANALYSIS")
print("-" * 80)

correlation_matrix = df[numerical_columns].corr()

plt.figure(figsize=(12, 8))
sns.heatmap(
    correlation_matrix,
    annot=True,
    fmt=".2f",
    cmap="coolwarm",
    center=0
)
plt.title("Correlation Heatmap")
plt.show()

exam_correlations = correlation_matrix["exam_score"].sort_values(
    ascending=False
)

print("\nCorrelation of Numerical Variables with Exam Score:")
display(exam_correlations.to_frame("Correlation with Exam Score"))

print("\n10. RELATIONSHIPS BETWEEN IMPORTANT FEATURES AND EXAM SCORE")
print("-" * 80)

important_numeric = [
    "study_hours_per_day",
    "attendance_percentage",
    "sleep_hours",
    "social_media_hours",
    "netflix_hours",
    "exercise_frequency",
    "mental_health_rating"
]

for col in important_numeric:
    if col in df.columns:
        plt.figure(figsize=(7, 5))
        sns.scatterplot(data=df, x=col, y="exam_score")
        sns.regplot(
            data=df,
            x=col,
            y="exam_score",
            scatter=False
        )
        plt.title(f"{col} vs Exam Score")
        plt.show()

print("\n11. CATEGORICAL VARIABLES VS EXAM SCORE")
print("-" * 80)

categorical_for_analysis = [
    "gender",
    "part_time_job",
    "diet_quality",
    "parental_education_level",
    "internet_quality",
    "extracurricular_participation"
]

for col in categorical_for_analysis:
    if col in df.columns:
        plt.figure(figsize=(9, 5))
        sns.boxplot(data=df, x=col, y="exam_score")
        plt.title(f"{col} vs Exam Score")
        plt.xticks(rotation=30)
        plt.show()

        print(f"\nAverage Exam Score by {col}:")
        display(
            df.groupby(col, dropna=False)["exam_score"]
            .agg(["mean", "median", "count"])
            .sort_values("mean", ascending=False)
        )

print("\n12. EXAM SCORE DISTRIBUTION")
print("-" * 80)

plt.figure(figsize=(9, 5))
sns.histplot(df["exam_score"], kde=True)
plt.axvline(50, linestyle="--", label="Passing Mark = 50")
plt.title("Distribution of Exam Scores")
plt.xlabel("Exam Score")
plt.ylabel("Frequency")
plt.legend()
plt.show()

print("Mean Exam Score:", df["exam_score"].mean())
print("Median Exam Score:", df["exam_score"].median())
print("Minimum Exam Score:", df["exam_score"].min())
print("Maximum Exam Score:", df["exam_score"].max())

print("\n13. FEATURE PREPARATION FOR MACHINE LEARNING")
print("-" * 80)

df_ml = df.copy()

df_ml["pass_fail"] = np.where(
    df_ml["exam_score"] >= 50,
    1,
    0
)

features = [
    "age",
    "gender",
    "study_hours_per_day",
    "social_media_hours",
    "netflix_hours",
    "part_time_job",
    "attendance_percentage",
    "sleep_hours",
    "diet_quality",
    "exercise_frequency",
    "parental_education_level",
    "internet_quality",
    "mental_health_rating",
    "extracurricular_participation"
]

X = df_ml[features]
y_regression = df_ml["exam_score"]
y_classification = df_ml["pass_fail"]

numeric_features = X.select_dtypes(include=np.number).columns.tolist()
categorical_features = X.select_dtypes(include="object").columns.tolist()

print("Features Used:")
print(features)

print("\nNumeric Features:")
print(numeric_features)

print("\nCategorical Features:")
print(categorical_features)

numeric_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="median")),
    ("scaler", StandardScaler())
])

categorical_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("encoder", OneHotEncoder(handle_unknown="ignore"))
])

preprocessor = ColumnTransformer([
    ("num", numeric_pipeline, numeric_features),
    ("cat", categorical_pipeline, categorical_features)
])

X_train, X_test, y_train_reg, y_test_reg = train_test_split(
    X,
    y_regression,
    test_size=0.20,
    random_state=42
)

X_train_cls, X_test_cls, y_train_cls, y_test_cls = train_test_split(
    X,
    y_classification,
    test_size=0.20,
    random_state=42,
    stratify=y_classification
)

print("\nTraining Samples:", X_train.shape[0])
print("Testing Samples:", X_test.shape[0])

print("\n14. LINEAR REGRESSION MODEL")
print("-" * 80)

linear_regression_model = Pipeline([
    ("preprocessor", preprocessor),
    ("model", LinearRegression())
])

linear_regression_model.fit(X_train, y_train_reg)

train_predictions_reg = linear_regression_model.predict(X_train)
test_predictions_reg = linear_regression_model.predict(X_test)

train_mae = mean_absolute_error(y_train_reg, train_predictions_reg)
test_mae = mean_absolute_error(y_test_reg, test_predictions_reg)

train_mse = mean_squared_error(y_train_reg, train_predictions_reg)
test_mse = mean_squared_error(y_test_reg, test_predictions_reg)

train_rmse = np.sqrt(train_mse)
test_rmse = np.sqrt(test_mse)

train_r2 = r2_score(y_train_reg, train_predictions_reg)
test_r2 = r2_score(y_test_reg, test_predictions_reg)

regression_results = pd.DataFrame({
    "Metric": ["MAE", "MSE", "RMSE", "R² Score"],
    "Training": [
        train_mae,
        train_mse,
        train_rmse,
        train_r2
    ],
    "Testing": [
        test_mae,
        test_mse,
        test_rmse,
        test_r2
    ]
})

print("Regression Model Evaluation:")
display(regression_results)

print("\n15. ACTUAL VS PREDICTED EXAM SCORES")
print("-" * 80)

plt.figure(figsize=(8, 6))
sns.scatterplot(
    x=y_test_reg,
    y=test_predictions_reg
)
plt.xlabel("Actual Exam Score")
plt.ylabel("Predicted Exam Score")
plt.title("Actual vs Predicted Exam Scores")

minimum_score = min(y_test_reg.min(), test_predictions_reg.min())
maximum_score = max(y_test_reg.max(), test_predictions_reg.max())

plt.plot(
    [minimum_score, maximum_score],
    [minimum_score, maximum_score],
    linestyle="--"
)

plt.show()

print("\n16. LINEAR REGRESSION COEFFICIENT ANALYSIS")
print("-" * 80)

reg_preprocessor = linear_regression_model.named_steps["preprocessor"]
reg_model = linear_regression_model.named_steps["model"]

feature_names = reg_preprocessor.get_feature_names_out()
coefficients = reg_model.coef_

coefficient_table = pd.DataFrame({
    "Feature": feature_names,
    "Coefficient": coefficients,
    "Absolute Importance": np.abs(coefficients)
}).sort_values(
    "Absolute Importance",
    ascending=False
)

display(coefficient_table.head(15))

print("\n17. CLASSIFICATION TARGET CREATION")
print("-" * 80)

print("Pass = Exam Score >= 50")
print("Fail = Exam Score < 50")

class_distribution = df_ml["pass_fail"].value_counts().sort_index()

print("\nClass Distribution:")
print(
    class_distribution.rename({
        0: "Fail",
        1: "Pass"
    })
)

plt.figure(figsize=(7, 5))
sns.countplot(
    x=df_ml["pass_fail"].map({
        0: "Fail",
        1: "Pass"
    })
)
plt.title("Pass vs Fail Distribution")
plt.xlabel("Result")
plt.ylabel("Number of Students")
plt.show()

print("\n18. LOGISTIC REGRESSION CLASSIFICATION MODEL")
print("-" * 80)

classification_model = Pipeline([
    ("preprocessor", preprocessor),
    (
        "model",
        LogisticRegression(
            max_iter=2000,
            random_state=42
        )
    )
])

classification_model.fit(X_train_cls, y_train_cls)

train_predictions_cls = classification_model.predict(X_train_cls)
test_predictions_cls = classification_model.predict(X_test_cls)

train_accuracy = accuracy_score(
    y_train_cls,
    train_predictions_cls
)

test_accuracy = accuracy_score(
    y_test_cls,
    test_predictions_cls
)

train_precision = precision_score(
    y_train_cls,
    train_predictions_cls,
    zero_division=0
)

test_precision = precision_score(
    y_test_cls,
    test_predictions_cls,
    zero_division=0
)

train_recall = recall_score(
    y_train_cls,
    train_predictions_cls,
    zero_division=0
)

test_recall = recall_score(
    y_test_cls,
    test_predictions_cls,
    zero_division=0
)

train_f1 = f1_score(
    y_train_cls,
    train_predictions_cls,
    zero_division=0
)

test_f1 = f1_score(
    y_test_cls,
    test_predictions_cls,
    zero_division=0
)

classification_results = pd.DataFrame({
    "Metric": [
        "Accuracy",
        "Precision",
        "Recall",
        "F1 Score"
    ],
    "Training": [
        train_accuracy,
        train_precision,
        train_recall,
        train_f1
    ],
    "Testing": [
        test_accuracy,
        test_precision,
        test_recall,
        test_f1
    ]
})

print("Classification Model Evaluation:")
display(classification_results)

print("\n19. CONFUSION MATRIX")
print("-" * 80)

cm = confusion_matrix(
    y_test_cls,
    test_predictions_cls
)

plt.figure(figsize=(7, 5))
sns.heatmap(
    cm,
    annot=True,
    fmt="d",
    cmap="Blues",
    xticklabels=["Fail", "Pass"],
    yticklabels=["Fail", "Pass"]
)
plt.xlabel("Predicted")
plt.ylabel("Actual")
plt.title("Confusion Matrix - Logistic Regression")
plt.show()

print("Confusion Matrix:")
print(cm)

print("\nClassification Report:")
print(
    classification_report(
        y_test_cls,
        test_predictions_cls,
        target_names=["Fail", "Pass"],
        zero_division=0
    )
)

print("\n20. TRAINING VS TESTING PERFORMANCE")
print("-" * 80)

print("REGRESSION")
print(f"Training R²  : {train_r2:.4f}")
print(f"Testing R²   : {test_r2:.4f}")
print(f"Training RMSE: {train_rmse:.4f}")
print(f"Testing RMSE : {test_rmse:.4f}")

print("\nCLASSIFICATION")
print(f"Training Accuracy: {train_accuracy:.4f}")
print(f"Testing Accuracy : {test_accuracy:.4f}")
print(f"Training F1 Score: {train_f1:.4f}")
print(f"Testing F1 Score : {test_f1:.4f}")

regression_r2_gap = train_r2 - test_r2
classification_accuracy_gap = train_accuracy - test_accuracy

print("\n21. OVERFITTING / UNDERFITTING ANALYSIS")
print("-" * 80)

if regression_r2_gap > 0.15:
    regression_fit = "The regression model shows signs of overfitting because training R² is considerably higher than testing R²."
elif train_r2 < 0.50 and test_r2 < 0.50:
    regression_fit = "The regression model may be underfitting because both training and testing R² scores are relatively low."
else:
    regression_fit = "The regression model does not show strong evidence of overfitting or underfitting."

if classification_accuracy_gap > 0.10:
    classification_fit = "The classification model shows signs of overfitting because training accuracy is considerably higher than testing accuracy."
elif train_accuracy < 0.65 and test_accuracy < 0.65:
    classification_fit = "The classification model may be underfitting because both training and testing accuracy are relatively low."
else:
    classification_fit = "The classification model does not show strong evidence of overfitting or underfitting."

print("Regression:", regression_fit)
print("Classification:", classification_fit)

print("\n22. EDA FINDINGS")
print("-" * 80)

numeric_corr = exam_correlations.drop("exam_score")

strongest_positive_feature = numeric_corr.idxmax()
strongest_positive_value = numeric_corr.max()

strongest_negative_feature = numeric_corr.idxmin()
strongest_negative_value = numeric_corr.min()

print(
    f"Strongest positive numerical relationship with exam_score: "
    f"{strongest_positive_feature} "
    f"(correlation = {strongest_positive_value:.3f})"
)

print(
    f"Strongest negative numerical relationship with exam_score: "
    f"{strongest_negative_feature} "
    f"(correlation = {strongest_negative_value:.3f})"
)

print(
    f"Average exam score: {df['exam_score'].mean():.2f}"
)

print(
    f"Median exam score: {df['exam_score'].median():.2f}"
)

pass_percentage = (
    df_ml["pass_fail"].mean() * 100
)

print(
    f"Pass percentage: {pass_percentage:.2f}%"
)

print(
    f"Fail percentage: {100 - pass_percentage:.2f}%"
)

print("\n23. TOP FEATURES FROM LINEAR REGRESSION")
print("-" * 80)

display(
    coefficient_table[
        ["Feature", "Coefficient"]
    ].head(10)
)

print("\n24. FIVE MEANINGFUL INSIGHTS ABOUT STUDENT PERFORMANCE")
print("-" * 80)

top_positive_features = coefficient_table[
    coefficient_table["Coefficient"] > 0
].head(3)

top_negative_features = coefficient_table[
    coefficient_table["Coefficient"] < 0
].head(3)

print(
    f"1. {strongest_positive_feature} has the strongest positive "
    f"numerical correlation with exam_score, with a correlation of "
    f"{strongest_positive_value:.3f}."
)

print(
    f"2. {strongest_negative_feature} has the strongest negative "
    f"numerical correlation with exam_score, with a correlation of "
    f"{strongest_negative_value:.3f}. This suggests that higher values "
    f"of this factor are associated with lower exam performance."
)

print(
    f"3. The average exam score is {df['exam_score'].mean():.2f}, "
    f"while the median is {df['exam_score'].median():.2f}. "
    f"This provides an overall picture of student academic performance."
)

print(
    f"4. {pass_percentage:.2f}% of students achieved a passing score "
    f"of at least 50 marks, while {100-pass_percentage:.2f}% were below "
    f"the passing threshold."
)

if len(top_positive_features) > 0:
    positive_names = ", ".join(
        top_positive_features["Feature"]
        .str.replace("num__", "", regex=False)
        .str.replace("cat__", "", regex=False)
        .head(3)
    )
    print(
        f"5. The Linear Regression coefficient analysis identifies "
        f"{positive_names} among the important positive contributors "
        f"to predicted exam performance. These relationships should "
        f"be interpreted as associations rather than proof of causation."
    )
else:
    print(
        "5. The coefficient analysis indicates that the selected "
        "features do not show strong positive linear contributions "
        "within this fitted model."
    )

print("\n25. FINAL MODEL SUMMARY")
print("-" * 80)

print(
    f"Linear Regression Testing R²: {test_r2:.4f}"
)

print(
    f"Linear Regression Testing RMSE: {test_rmse:.4f}"
)

print(
    f"Linear Regression Testing MAE: {test_mae:.4f}"
)

print(
    f"Logistic Regression Testing Accuracy: {test_accuracy:.4f}"
)

print(
    f"Logistic Regression Testing Precision: {test_precision:.4f}"
)

print(
    f"Logistic Regression Testing Recall: {test_recall:.4f}"
)

print(
    f"Logistic Regression Testing F1 Score: {test_f1:.4f}"
)

print("\nAssignment completed successfully.")
```
