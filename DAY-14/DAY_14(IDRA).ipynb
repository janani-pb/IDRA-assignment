```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from google.colab import files

sns.set_theme(style="whitegrid")
plt.rcParams["figure.figsize"] = (10, 6)

print("=" * 70)
print("FOOD DELIVERY BUSINESS PERFORMANCE - VISUALIZATION PORTFOLIO")
print("=" * 70)

uploaded = files.upload()
file_name = list(uploaded.keys())[0]

df = pd.read_csv(file_name)

df["Date"] = pd.to_datetime(df["Date"])

print("\nDATASET LOADED SUCCESSFULLY")
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])

print("\nCOLUMN NAMES")
print(df.columns.tolist())

print("\nFIRST FIVE RECORDS")
display(df.head())

print("\nDATASET INFORMATION")
df.info()

print("\nMISSING VALUES")
display(df.isnull().sum().to_frame("Missing Values"))

print("\nDESCRIPTIVE STATISTICS")
display(df.describe().T)


# ============================================================
# 1. LINE PLOT - ORDERS OVER TIME
# ============================================================

daily_orders = df.groupby("Date")["Orders"].sum().sort_index()

plt.figure(figsize=(14, 6))
plt.plot(
    daily_orders.index,
    daily_orders.values,
    marker="o",
    linewidth=2,
    label="Orders"
)

plt.title("Orders Over Time", fontsize=16, fontweight="bold")
plt.xlabel("Date")
plt.ylabel("Number of Orders")
plt.legend()
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()

print("""
INTERPRETATION:
The line plot shows how the number of food delivery orders changes over
time. Peaks indicate periods of higher demand, while lower points indicate
periods of reduced order activity. This helps identify demand fluctuations
and potential high-demand periods.
""")


# ============================================================
# 2. LINE PLOT - REVENUE OVER TIME
# ============================================================

daily_revenue = df.groupby("Date")["Revenue"].sum().sort_index()

plt.figure(figsize=(14, 6))
plt.plot(
    daily_revenue.index,
    daily_revenue.values,
    marker="o",
    linewidth=2,
    label="Revenue"
)

plt.title("Revenue Trend Over Time", fontsize=16, fontweight="bold")
plt.xlabel("Date")
plt.ylabel("Revenue")
plt.legend()
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()

print("""
INTERPRETATION:
Revenue changes across different dates. Higher revenue periods may be
associated with increased order volume and/or higher average order values.
The chart helps identify strong and weak revenue periods.
""")


# ============================================================
# 3. BAR CHART - REVENUE BY CITY
# ============================================================

city_revenue = (
    df.groupby("City")["Revenue"]
    .sum()
    .sort_values(ascending=False)
)

plt.figure(figsize=(12, 6))
sns.barplot(
    x=city_revenue.index,
    y=city_revenue.values
)

plt.title("Total Revenue by City", fontsize=16, fontweight="bold")
plt.xlabel("City")
plt.ylabel("Total Revenue")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()

print("""
INTERPRETATION:
The bar chart compares revenue generated by different cities. Cities with
higher bars contribute more revenue and may represent stronger markets
for the food delivery business.
""")


# ============================================================
# 4. BAR CHART - ORDERS BY CUISINE
# ============================================================

cuisine_orders = (
    df.groupby("Cuisine")["Orders"]
    .sum()
    .sort_values(ascending=False)
)

plt.figure(figsize=(12, 6))
sns.barplot(
    x=cuisine_orders.index,
    y=cuisine_orders.values
)

plt.title("Total Orders by Cuisine", fontsize=16, fontweight="bold")
plt.xlabel("Cuisine")
plt.ylabel("Total Orders")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()

print("""
INTERPRETATION:
The chart shows which cuisines receive the highest number of orders.
Higher-order cuisines can be considered popular customer choices and may
deserve greater promotional attention.
""")


# ============================================================
# 5. BAR CHART - REVENUE BY ORDER CHANNEL
# ============================================================

channel_revenue = (
    df.groupby("Order_Channel")["Revenue"]
    .sum()
    .sort_values(ascending=False)
)

plt.figure(figsize=(10, 6))
sns.barplot(
    x=channel_revenue.index,
    y=channel_revenue.values
)

plt.title("Revenue by Order Channel", fontsize=16, fontweight="bold")
plt.xlabel("Order Channel")
plt.ylabel("Total Revenue")
plt.xticks(rotation=30)
plt.tight_layout()
plt.show()

print("""
INTERPRETATION:
The visualization compares revenue generated through different order
channels. The leading channel contributes the highest revenue and can be
important for future digital marketing and operational strategies.
""")


# ============================================================
# 6. SCATTER PLOT - MARKETING SPEND VS REVENUE
# ============================================================

plt.figure(figsize=(10, 6))

sns.scatterplot(
    data=df,
    x="Marketing_Spend",
    y="Revenue",
    hue="City",
    s=80
)

plt.title("Marketing Spend vs Revenue", fontsize=16, fontweight="bold")
plt.xlabel("Marketing Spend")
plt.ylabel("Revenue")
plt.legend(bbox_to_anchor=(1.05, 1), loc="upper left")
plt.tight_layout()
plt.show()

marketing_corr = df["Marketing_Spend"].corr(df["Revenue"])

print(f"""
INTERPRETATION:
The scatter plot investigates the relationship between marketing spending
and revenue. The correlation between marketing spend and revenue is
{marketing_corr:.2f}. A positive value indicates that higher marketing
spending tends to be associated with higher revenue, while a value close
to zero indicates a weak relationship.
""")


# ============================================================
# 7. SCATTER PLOT - DELIVERY TIME VS CUSTOMER RATING
# ============================================================

plt.figure(figsize=(10, 6))

sns.scatterplot(
    data=df,
    x="Avg_Delivery_Minutes",
    y="Customer_Rating",
    hue="Weather",
    s=80
)

plt.title(
    "Delivery Time vs Customer Rating",
    fontsize=16,
    fontweight="bold"
)

plt.xlabel("Average Delivery Time (Minutes)")
plt.ylabel("Customer Rating")

plt.tight_layout()
plt.show()

delivery_rating_corr = df["Avg_Delivery_Minutes"].corr(
    df["Customer_Rating"]
)

print(f"""
INTERPRETATION:
The chart examines whether delivery time is associated with customer
ratings. The correlation is {delivery_rating_corr:.2f}. A negative
correlation would indicate that longer delivery times tend to be linked
with lower ratings.
""")


# ============================================================
# 8. HISTOGRAM - ORDER DISTRIBUTION
# ============================================================

plt.figure(figsize=(10, 6))

sns.histplot(
    data=df,
    x="Orders",
    bins=20,
    kde=True
)

plt.title("Distribution of Orders", fontsize=16, fontweight="bold")
plt.xlabel("Number of Orders")
plt.ylabel("Frequency")
plt.tight_layout()
plt.show()

print("""
INTERPRETATION:
The histogram shows how order volumes are distributed across the dataset.
It helps identify the typical order range and whether unusually high or
low order volumes occur.
""")


# ============================================================
# 9. HISTOGRAM - REVENUE DISTRIBUTION
# ============================================================

plt.figure(figsize=(10, 6))

sns.histplot(
    data=df,
    x="Revenue",
    bins=20,
    kde=True
)

plt.title("Distribution of Revenue", fontsize=16, fontweight="bold")
plt.xlabel("Revenue")
plt.ylabel("Frequency")
plt.tight_layout()
plt.show()

print("""
INTERPRETATION:
The revenue histogram shows the distribution of revenue values across
order batches. It helps identify the common revenue range and possible
extreme revenue observations.
""")


# ============================================================
# 10. BOX PLOT - REVENUE BY CITY
# ============================================================

plt.figure(figsize=(12, 6))

sns.boxplot(
    data=df,
    x="City",
    y="Revenue"
)

plt.title("Revenue Distribution Across Cities", fontsize=16, fontweight="bold")
plt.xlabel("City")
plt.ylabel("Revenue")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()

print("""
INTERPRETATION:
The box plot compares revenue distributions across cities. Differences
in median revenue, spread, and outliers help identify cities with
consistently strong or highly variable revenue performance.
""")


# ============================================================
# 11. BOX PLOT - DELIVERY TIME BY WEATHER
# ============================================================

plt.figure(figsize=(10, 6))

sns.boxplot(
    data=df,
    x="Weather",
    y="Avg_Delivery_Minutes"
)

plt.title(
    "Delivery Time Distribution by Weather",
    fontsize=16,
    fontweight="bold"
)

plt.xlabel("Weather")
plt.ylabel("Average Delivery Time (Minutes)")
plt.tight_layout()
plt.show()

print("""
INTERPRETATION:
The box plot compares delivery times under different weather conditions.
Weather categories with higher median delivery times may create greater
operational challenges for food delivery services.
""")


# ============================================================
# 12. VIOLIN PLOT - CUSTOMER RATINGS BY CUISINE
# ============================================================

plt.figure(figsize=(12, 6))

sns.violinplot(
    data=df,
    x="Cuisine",
    y="Customer_Rating"
)

plt.title(
    "Customer Rating Distribution by Cuisine",
    fontsize=16,
    fontweight="bold"
)

plt.xlabel("Cuisine")
plt.ylabel("Customer Rating")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()

print("""
INTERPRETATION:
The violin plot shows the distribution and density of customer ratings
for each cuisine. It helps identify cuisines with higher typical ratings
and cuisines with wider variation in customer satisfaction.
""")


# ============================================================
# 13. VIOLIN PLOT - REVENUE BY ORDER CHANNEL
# ============================================================

plt.figure(figsize=(10, 6))

sns.violinplot(
    data=df,
    x="Order_Channel",
    y="Revenue"
)

plt.title(
    "Revenue Distribution by Order Channel",
    fontsize=16,
    fontweight="bold"
)

plt.xlabel("Order Channel")
plt.ylabel("Revenue")
plt.xticks(rotation=30)
plt.tight_layout()
plt.show()

print("""
INTERPRETATION:
The violin plot compares revenue distributions across order channels.
It shows both the typical revenue levels and the variation within each
channel.
""")


# ============================================================
# 14. COUNT PLOT - ORDERS BY WEATHER
# ============================================================

plt.figure(figsize=(10, 6))

sns.countplot(
    data=df,
    x="Weather"
)

plt.title(
    "Number of Order Batches by Weather",
    fontsize=16,
    fontweight="bold"
)

plt.xlabel("Weather")
plt.ylabel("Count")
plt.tight_layout()
plt.show()

print("""
INTERPRETATION:
The count plot shows how frequently each weather condition appears in the
dataset. This provides context for comparing operational performance under
different weather conditions.
""")


# ============================================================
# 15. COUNT PLOT - ORDER CHANNEL
# ============================================================

plt.figure(figsize=(10, 6))

sns.countplot(
    data=df,
    x="Order_Channel"
)

plt.title(
    "Number of Order Batches by Order Channel",
    fontsize=16,
    fontweight="bold"
)

plt.xlabel("Order Channel")
plt.ylabel("Count")
plt.xticks(rotation=30)
plt.tight_layout()
plt.show()

print("""
INTERPRETATION:
The count plot shows the frequency of different order channels. A higher
count indicates that the channel is more frequently represented in the
dataset.
""")


# ============================================================
# 16. WEATHER VS AVERAGE REVENUE
# ============================================================

weather_revenue = (
    df.groupby("Weather")["Revenue"]
    .mean()
    .sort_values(ascending=False)
)

plt.figure(figsize=(10, 6))

sns.barplot(
    x=weather_revenue.index,
    y=weather_revenue.values
)

plt.title(
    "Average Revenue by Weather Condition",
    fontsize=16,
    fontweight="bold"
)

plt.xlabel("Weather")
plt.ylabel("Average Revenue")
plt.tight_layout()
plt.show()

print("""
INTERPRETATION:
Average revenue differs across weather conditions. This can help the
business understand whether weather conditions are associated with changes
in customer spending and delivery demand.
""")


# ============================================================
# 17. ORDER CHANNEL VS AVERAGE ORDERS
# ============================================================

channel_orders = (
    df.groupby("Order_Channel")["Orders"]
    .mean()
    .sort_values(ascending=False)
)

plt.figure(figsize=(10, 6))

sns.barplot(
    x=channel_orders.index,
    y=channel_orders.values
)

plt.title(
    "Average Orders by Order Channel",
    fontsize=16,
    fontweight="bold"
)

plt.xlabel("Order Channel")
plt.ylabel("Average Orders")
plt.xticks(rotation=30)
plt.tight_layout()
plt.show()

print("""
INTERPRETATION:
The chart compares average order volume across order channels. The channel
with the highest average orders represents the strongest source of order
volume in the dataset.
""")


# ============================================================
# 18. CITY VS AVERAGE CUSTOMER RATING
# ============================================================

city_rating = (
    df.groupby("City")["Customer_Rating"]
    .mean()
    .sort_values(ascending=False)
)

plt.figure(figsize=(12, 6))

sns.barplot(
    x=city_rating.index,
    y=city_rating.values
)

plt.title(
    "Average Customer Rating by City",
    fontsize=16,
    fontweight="bold"
)

plt.xlabel("City")
plt.ylabel("Average Customer Rating")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()

print("""
INTERPRETATION:
The chart compares customer satisfaction across cities. Cities with
higher average ratings demonstrate stronger customer satisfaction.
""")


# ============================================================
# 19. CORRELATION HEATMAP
# ============================================================

numeric_columns = [
    "Orders",
    "Average_Order_Value",
    "Revenue",
    "Marketing_Spend",
    "Discounts",
    "Avg_Delivery_Minutes",
    "Customer_Rating",
    "Repeat_Customer_Percent"
]

correlation = df[numeric_columns].corr()

plt.figure(figsize=(12, 9))

sns.heatmap(
    correlation,
    annot=True,
    fmt=".2f",
    cmap="coolwarm",
    linewidths=0.5,
    square=True
)

plt.title(
    "Correlation Heatmap of Food Delivery Performance Variables",
    fontsize=16,
    fontweight="bold"
)

plt.tight_layout()
plt.show()

print("""
INTERPRETATION:
The correlation heatmap summarizes relationships between numerical
variables. Strong positive values indicate variables that increase
together, while strong negative values indicate an inverse relationship.
Values close to zero indicate weak linear relationships.
""")


# ============================================================
# 20. REVENUE VS ORDERS
# ============================================================

plt.figure(figsize=(10, 6))

sns.scatterplot(
    data=df,
    x="Orders",
    y="Revenue",
    hue="Cuisine",
    s=80
)

plt.title(
    "Orders vs Revenue by Cuisine",
    fontsize=16,
    fontweight="bold"
)

plt.xlabel("Orders")
plt.ylabel("Revenue")
plt.legend(bbox_to_anchor=(1.05, 1), loc="upper left")
plt.tight_layout()
plt.show()

orders_revenue_corr = df["Orders"].corr(df["Revenue"])

print(f"""
INTERPRETATION:
The scatter plot examines whether higher order volumes are associated
with higher revenue. The correlation between orders and revenue is
{orders_revenue_corr:.2f}.
""")


# ============================================================
# 21. REPEAT CUSTOMERS VS CUSTOMER RATING
# ============================================================

plt.figure(figsize=(10, 6))

sns.scatterplot(
    data=df,
    x="Customer_Rating",
    y="Repeat_Customer_Percent",
    hue="City",
    s=80
)

plt.title(
    "Customer Rating vs Repeat Customer Percentage",
    fontsize=16,
    fontweight="bold"
)

plt.xlabel("Customer Rating")
plt.ylabel("Repeat Customer Percentage")
plt.tight_layout()
plt.show()

repeat_rating_corr = df["Customer_Rating"].corr(
    df["Repeat_Customer_Percent"]
)

print(f"""
INTERPRETATION:
The chart investigates whether customer satisfaction is associated with
repeat customers. The correlation between rating and repeat customer
percentage is {repeat_rating_corr:.2f}.
""")


# ============================================================
# 22. TOP CITIES BY REVENUE
# ============================================================

top_cities = (
    df.groupby("City")["Revenue"]
    .sum()
    .sort_values(ascending=False)
)

print("\nTOP CITIES BY TOTAL REVENUE")
display(top_cities.to_frame("Total Revenue"))


# ============================================================
# 23. TOP CUISINES BY REVENUE
# ============================================================

top_cuisines = (
    df.groupby("Cuisine")["Revenue"]
    .sum()
    .sort_values(ascending=False)
)

print("\nTOP CUISINES BY TOTAL REVENUE")
display(top_cuisines.to_frame("Total Revenue"))


# ============================================================
# 24. TOP ORDER CHANNEL BY REVENUE
# ============================================================

top_channel = (
    df.groupby("Order_Channel")["Revenue"]
    .sum()
    .sort_values(ascending=False)
)

print("\nREVENUE BY ORDER CHANNEL")
display(top_channel.to_frame("Total Revenue"))


# ============================================================
# 25. FINAL DATA-DRIVEN FINDINGS
# ============================================================

highest_revenue_city = top_cities.idxmax()
highest_revenue_city_value = top_cities.max()

highest_revenue_cuisine = top_cuisines.idxmax()
highest_revenue_cuisine_value = top_cuisines.max()

highest_revenue_channel = top_channel.idxmax()
highest_revenue_channel_value = top_channel.max()

highest_orders_city = (
    df.groupby("City")["Orders"]
    .sum()
    .sort_values(ascending=False)
    .idxmax()
)

highest_rating_city = (
    df.groupby("City")["Customer_Rating"]
    .mean()
    .sort_values(ascending=False)
    .idxmax()
)

fastest_weather = (
    df.groupby("Weather")["Avg_Delivery_Minutes"]
    .mean()
    .sort_values()
    .idxmin()
)

highest_repeat_city = (
    df.groupby("City")["Repeat_Customer_Percent"]
    .mean()
    .sort_values(ascending=False)
    .idxmax()
)

print("\n" + "=" * 70)
print("FINAL SUMMARY OF IMPORTANT FINDINGS")
print("=" * 70)

print(f"""
1. {highest_revenue_city} generates the highest total revenue among the
   cities, with approximately {highest_revenue_city_value:,.2f} in revenue.

2. {highest_revenue_cuisine} is the highest-revenue cuisine, indicating
   strong customer demand for this cuisine category.

3. {highest_revenue_channel} generates the highest total revenue among
   the available order channels.

4. {highest_orders_city} records the highest total order volume among
   the cities, making it an important market for the food delivery
   business.

5. {highest_rating_city} has the highest average customer rating,
   indicating comparatively strong customer satisfaction.

6. {fastest_weather} weather has the lowest average delivery time among
   the weather categories in the dataset.

7. The relationship between marketing spend and revenue can be evaluated
   using the scatter plot and correlation analysis, helping management
   understand whether additional marketing investment is associated with
   increased revenue.

8. The correlation heatmap provides an overall view of relationships
   among orders, average order value, revenue, marketing spend, discounts,
   delivery time, customer rating, and repeat customer percentage.
""")

print("=" * 70)
print("VISUALIZATION PORTFOLIO COMPLETED")
print("=" * 70)
```
