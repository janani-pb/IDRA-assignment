```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

df = pd.read_csv("Day10_Flight_Operations_Dataset.csv")
print("FLIGHT OPERATIONS DATA ANALYSIS - DAY 10")

# 1. LOAD AND UNDERSTAND DATASET
print("\n1. FIRST 5 ROWS")
display(df.head())

print("\n2. LAST 5 ROWS")
display(df.tail())

print("\n3. RANDOM 5 ROWS")
display(df.sample(5))

print("\n4. DATASET SHAPE")
print("Number of Rows:", df.shape[0])
print("Number of Columns:", df.shape[1])

print("\n5. COLUMN NAMES")
print(df.columns.tolist())

print("\n6. DATA TYPES")
print(df.dtypes)

print("\n7. DATASET INFORMATION")
df.info()

print("\n8. DESCRIPTIVE STATISTICS")
display(df.describe())

print("\n9. MISSING VALUES")
display(df.isnull().sum())

print("\n10. DUPLICATE ROWS")
print(df.duplicated().sum())


# 2. DATETIME TRANSFORMATION
df["Flight_Date"] = pd.to_datetime(df["Flight_Date"])

df["Flight_Year"] = df["Flight_Date"].dt.year
df["Flight_Month"] = df["Flight_Date"].dt.month
df["Flight_Month_Name"] = df["Flight_Date"].dt.month_name()
df["Flight_Day"] = df["Flight_Date"].dt.day
df["Flight_Day_Name"] = df["Flight_Date"].dt.day_name()

print("\n11. DATE TIME TRANSFORMATION")
display(
    df[
        [
            "Flight_Date",
            "Flight_Year",
            "Flight_Month",
            "Flight_Month_Name",
            "Flight_Day",
            "Flight_Day_Name"
        ]
    ].head(10)
)


# 3. DATA SELECTION
print("\n12. SELECT FLIGHT, AIRLINE AND STATUS")
display(df[["Flight_ID", "Airline", "Flight_Status"]].head(10))

print("\n13. SELECT AIRLINE, ROUTE AND PASSENGERS")
display(
    df[
        [
            "Airline",
            "Origin",
            "Destination",
            "Passengers"
        ]
    ].head(10)
)

print("\n14. SELECT TICKET PRICE AND SATISFACTION")
display(
    df[
        [
            "Average_Ticket_Price",
            "Passenger_Satisfaction"
        ]
    ].head(10)
)


# 4. FILTERING
print("\n15. FLIGHTS WITH MORE THAN 150 PASSENGERS")
display(df[df["Passengers"] > 150])

print("\n16. FLIGHTS WITH DELAY GREATER THAN 60 MINUTES")
display(df[df["Delay_Minutes"] > 60])

print("\n17. FLIGHTS WITH SATISFACTION GREATER THAN OR EQUAL TO 4")
display(df[df["Passenger_Satisfaction"] >= 4])

print("\n18. FLIGHTS WITH TICKET PRICE GREATER THAN 15000")
display(df[df["Average_Ticket_Price"] > 15000])

print("\n19. FLIGHTS WITH MORE THAN 150 PASSENGERS AND DELAY GREATER THAN 30 MINUTES")
display(
    df[
        (df["Passengers"] > 150) &
        (df["Delay_Minutes"] > 30)
    ]
)

print("\n20. DELAYED FLIGHTS")
display(df[df["Flight_Status"] == "Delayed"])


# 5. SORTING
print("\n21. TOP 10 FLIGHTS BY PASSENGERS")
display(
    df.sort_values(
        "Passengers",
        ascending=False
    ).head(10)
)

print("\n22. TOP 10 FLIGHTS BY TICKET PRICE")
display(
    df.sort_values(
        "Average_Ticket_Price",
        ascending=False
    ).head(10)
)

print("\n23. TOP 10 MOST DELAYED FLIGHTS")
display(
    df.sort_values(
        "Delay_Minutes",
        ascending=False
    ).head(10)
)

print("\n24. HIGHEST PASSENGER SATISFACTION")
display(
    df.sort_values(
        "Passenger_Satisfaction",
        ascending=False
    ).head(10)
)


# 6. OVERALL STATISTICS
total_passengers = df["Passengers"].sum()
average_passengers = df["Passengers"].mean()
maximum_passengers = df["Passengers"].max()
minimum_passengers = df["Passengers"].min()

average_delay = df["Delay_Minutes"].mean()
maximum_delay = df["Delay_Minutes"].max()

average_ticket_price = df["Average_Ticket_Price"].mean()
highest_ticket_price = df["Average_Ticket_Price"].max()
lowest_ticket_price = df["Average_Ticket_Price"].min()

average_satisfaction = df["Passenger_Satisfaction"].mean()

print("\n25. OVERALL FLIGHT STATISTICS")
print("Total Flights:", len(df))
print("Total Passengers:", total_passengers)
print("Average Passengers per Flight:", round(average_passengers, 2))
print("Maximum Passengers:", maximum_passengers)
print("Minimum Passengers:", minimum_passengers)
print("Average Delay:", round(average_delay, 2), "minutes")
print("Maximum Delay:", round(maximum_delay, 2), "minutes")
print("Average Ticket Price:", round(average_ticket_price, 2))
print("Highest Ticket Price:", highest_ticket_price)
print("Lowest Ticket Price:", lowest_ticket_price)
print("Average Passenger Satisfaction:", round(average_satisfaction, 2))


# 7. AIRLINE ANALYSIS
airline_analysis = df.groupby("Airline").agg(
    Total_Flights=("Flight_ID", "count"),
    Total_Passengers=("Passengers", "sum"),
    Average_Passengers=("Passengers", "mean"),
    Average_Delay=("Delay_Minutes", "mean"),
    Average_Ticket_Price=("Average_Ticket_Price", "mean"),
    Average_Satisfaction=("Passenger_Satisfaction", "mean")
).sort_values(
    "Total_Passengers",
    ascending=False
)

print("\n26. AIRLINE ANALYSIS")
display(airline_analysis)


# 8. ORIGIN ANALYSIS
origin_analysis = df.groupby("Origin").agg(
    Total_Flights=("Flight_ID", "count"),
    Total_Passengers=("Passengers", "sum"),
    Average_Delay=("Delay_Minutes", "mean"),
    Average_Satisfaction=("Passenger_Satisfaction", "mean")
).sort_values(
    "Total_Passengers",
    ascending=False
)

print("\n27. ORIGIN ANALYSIS")
display(origin_analysis)


# 9. DESTINATION ANALYSIS
destination_analysis = df.groupby("Destination").agg(
    Total_Flights=("Flight_ID", "count"),
    Total_Passengers=("Passengers", "sum"),
    Average_Delay=("Delay_Minutes", "mean"),
    Average_Satisfaction=("Passenger_Satisfaction", "mean")
).sort_values(
    "Total_Passengers",
    ascending=False
)

print("\n28. DESTINATION ANALYSIS")
display(destination_analysis)


# 10. AIRCRAFT ANALYSIS
aircraft_analysis = df.groupby("Aircraft").agg(
    Total_Flights=("Flight_ID", "count"),
    Total_Passengers=("Passengers", "sum"),
    Average_Delay=("Delay_Minutes", "mean"),
    Average_Satisfaction=("Passenger_Satisfaction", "mean")
).sort_values(
    "Total_Passengers",
    ascending=False
)

print("\n29. AIRCRAFT ANALYSIS")
display(aircraft_analysis)


# 11. TRAVEL CLASS ANALYSIS
class_analysis = df.groupby("Travel_Class").agg(
    Total_Flights=("Flight_ID", "count"),
    Total_Passengers=("Passengers", "sum"),
    Average_Ticket_Price=("Average_Ticket_Price", "mean"),
    Average_Satisfaction=("Passenger_Satisfaction", "mean")
).sort_values(
    "Total_Passengers",
    ascending=False
)

print("\n30. TRAVEL CLASS ANALYSIS")
display(class_analysis)


# 12. FLIGHT STATUS ANALYSIS
status_analysis = df.groupby("Flight_Status").agg(
    Number_of_Flights=("Flight_ID", "count"),
    Total_Passengers=("Passengers", "sum"),
    Average_Delay=("Delay_Minutes", "mean"),
    Average_Satisfaction=("Passenger_Satisfaction", "mean")
).sort_values(
    "Number_of_Flights",
    ascending=False
)

print("\n31. FLIGHT STATUS ANALYSIS")
display(status_analysis)


# 13. WEATHER ANALYSIS
weather_analysis = df.groupby("Weather").agg(
    Number_of_Flights=("Flight_ID", "count"),
    Average_Delay=("Delay_Minutes", "mean"),
    Average_Satisfaction=("Passenger_Satisfaction", "mean")
).sort_values(
    "Average_Delay",
    ascending=False
)

print("\n32. WEATHER ANALYSIS")
display(weather_analysis)


# 14. BOOKING CHANNEL ANALYSIS
booking_analysis = df.groupby("Booking_Channel").agg(
    Number_of_Flights=("Flight_ID", "count"),
    Total_Passengers=("Passengers", "sum"),
    Average_Ticket_Price=("Average_Ticket_Price", "mean"),
    Average_Satisfaction=("Passenger_Satisfaction", "mean")
).sort_values(
    "Total_Passengers",
    ascending=False
)

print("\n33. BOOKING CHANNEL ANALYSIS")
display(booking_analysis)


# 15. MEAL PREFERENCE ANALYSIS
meal_analysis = df.groupby("Meal_Preference").agg(
    Number_of_Flights=("Flight_ID", "count"),
    Total_Passengers=("Passengers", "sum"),
    Average_Satisfaction=("Passenger_Satisfaction", "mean")
).sort_values(
    "Total_Passengers",
    ascending=False
)

print("\n34. MEAL PREFERENCE ANALYSIS")
display(meal_analysis)


# 16. MONTHLY ANALYSIS
monthly_analysis = df.groupby(
    ["Flight_Month", "Flight_Month_Name"]
).agg(
    Number_of_Flights=("Flight_ID", "count"),
    Total_Passengers=("Passengers", "sum"),
    Average_Delay=("Delay_Minutes", "mean"),
    Average_Satisfaction=("Passenger_Satisfaction", "mean")
).sort_index()

print("\n35. MONTHLY FLIGHT ANALYSIS")
display(monthly_analysis)


# 17. ROUTE ANALYSIS
route_analysis = df.groupby(
    ["Origin", "Destination"]
).agg(
    Number_of_Flights=("Flight_ID", "count"),
    Total_Passengers=("Passengers", "sum"),
    Average_Delay=("Delay_Minutes", "mean"),
    Average_Ticket_Price=("Average_Ticket_Price", "mean")
).sort_values(
    "Total_Passengers",
    ascending=False
)

print("\n36. TOP ROUTES")
display(route_analysis.head(10))


# 18. OCCUPANCY RATE
df["Occupancy_Rate"] = (
    df["Passengers"] / df["Seat_Capacity"] * 100
)

print("\n37. OCCUPANCY RATE")
display(
    df[
        [
            "Flight_ID",
            "Passengers",
            "Seat_Capacity",
            "Occupancy_Rate"
        ]
    ].head(10)
)

print("\n38. TOP 10 FLIGHTS BY OCCUPANCY RATE")
display(
    df.sort_values(
        "Occupancy_Rate",
        ascending=False
    ).head(10)
)


# 19. DELAY CATEGORY
df["Delay_Category"] = df["Delay_Minutes"].apply(
    lambda x:
    "On Time" if x == 0
    else "Minor Delay" if x <= 30
    else "Moderate Delay" if x <= 60
    else "Major Delay"
)

print("\n39. DELAY CATEGORY")
display(
    df[
        [
            "Flight_ID",
            "Delay_Minutes",
            "Delay_Category"
        ]
    ].head(10)
)


# 20. TICKET PRICE CATEGORY
df["Ticket_Price_Category"] = pd.cut(
    df["Average_Ticket_Price"],
    bins=[0, 5000, 10000, 15000, np.inf],
    labels=[
        "Low",
        "Medium",
        "High",
        "Very High"
    ]
)

print("\n40. TICKET PRICE CATEGORY")
display(
    df[
        [
            "Flight_ID",
            "Average_Ticket_Price",
            "Ticket_Price_Category"
        ]
    ].head(10)
)


# 21. HIGHEST AND LOWEST VALUES
highest_passenger_flight = df.loc[
    df["Passengers"].idxmax()
]

most_delayed_flight = df.loc[
    df["Delay_Minutes"].idxmax()
]

highest_price_flight = df.loc[
    df["Average_Ticket_Price"].idxmax()
]

highest_satisfaction_flight = df.loc[
    df["Passenger_Satisfaction"].idxmax()
]

print("\n41. HIGHEST PASSENGER FLIGHT")
display(highest_passenger_flight.to_frame().T)

print("\n42. MOST DELAYED FLIGHT")
display(most_delayed_flight.to_frame().T)

print("\n43. HIGHEST TICKET PRICE FLIGHT")
display(highest_price_flight.to_frame().T)

print("\n44. HIGHEST SATISFACTION FLIGHT")
display(highest_satisfaction_flight.to_frame().T)


# 22. VISUALIZATION - AIRLINE
print("\n45. AIRLINE PASSENGER CHART")

airline_analysis["Total_Passengers"].plot(
    kind="bar",
    figsize=(10, 5),
    title="Total Passengers by Airline"
)

plt.xlabel("Airline")
plt.ylabel("Total Passengers")
plt.xticks(rotation=45)
plt.show()


# 23. VISUALIZATION - FLIGHT STATUS
print("\n46. FLIGHT STATUS CHART")

status_analysis["Number_of_Flights"].plot(
    kind="bar",
    figsize=(8, 5),
    title="Number of Flights by Status"
)

plt.xlabel("Flight Status")
plt.ylabel("Number of Flights")
plt.xticks(rotation=45)
plt.show()


# 24. VISUALIZATION - WEATHER AND DELAY
print("\n47. WEATHER DELAY CHART")

weather_analysis["Average_Delay"].plot(
    kind="bar",
    figsize=(8, 5),
    title="Average Delay by Weather Condition"
)

plt.xlabel("Weather")
plt.ylabel("Average Delay (Minutes)")
plt.xticks(rotation=45)
plt.show()


# 25. VISUALIZATION - TRAVEL CLASS
print("\n48. TRAVEL CLASS PASSENGERS")

class_analysis["Total_Passengers"].plot(
    kind="bar",
    figsize=(8, 5),
    title="Total Passengers by Travel Class"
)

plt.xlabel("Travel Class")
plt.ylabel("Total Passengers")
plt.xticks(rotation=45)
plt.show()


# 26. FINAL OBSERVATIONS
best_airline = airline_analysis["Total_Passengers"].idxmax()
best_origin = origin_analysis["Total_Passengers"].idxmax()
best_destination = destination_analysis["Total_Passengers"].idxmax()
worst_weather = weather_analysis["Average_Delay"].idxmax()
best_class = class_analysis["Total_Passengers"].idxmax()
best_booking_channel = booking_analysis["Total_Passengers"].idxmax()
best_route = route_analysis.index[0]

print("\n" + "=" * 70)
print("FINAL OBSERVATIONS")
print("=" * 70)

print(f"1. The dataset contains {len(df)} flight records.")
print(
    f"2. A total of {total_passengers:,} passengers "
    f"were recorded across all flights."
)
print(
    f"3. {best_airline} carried the highest total number "
    f"of passengers."
)
print(
    f"4. {best_origin} was the origin with the highest "
    f"total passenger traffic."
)
print(
    f"5. {best_destination} was the destination with the "
    f"highest total passenger traffic."
)
print(
    f"6. The average flight delay was "
    f"{average_delay:.2f} minutes."
)
print(
    f"7. {worst_weather} weather had the highest average "
    f"flight delay."
)
print(
    f"8. {best_class} had the highest number of passengers "
    f"among travel classes."
)
print(
    f"9. The most common/highest passenger booking channel "
    f"was {best_booking_channel}."
)
print(
    f"10. The busiest route based on passenger count was "
    f"{best_route[0]} to {best_route[1]}."
)

print("\n" + "=" * 70)
print("ANALYSIS COMPLETED SUCCESSFULLY")
print("=" * 70)
```
