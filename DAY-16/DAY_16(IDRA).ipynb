```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

import os

file_path = "/mnt/data/Day16_Student_Wellbeing_Survey.csv"

if not os.path.exists(file_path):
    file_path = "Day16_Student_Wellbeing_Survey.csv"

df = pd.read_csv(file_path)



print("\n" + "=" * 80)
print("1. DATASET INSPECTION")
print("=" * 80)

print("\nDataset Shape:")
print(f"Rows    : {df.shape[0]}")
print(f"Columns : {df.shape[1]}")

print("\nFirst 5 Rows:")
display(df.head())

print("\nColumn Names:")
print(df.columns.tolist())

print("\nData Types:")
print(df.dtypes)

print("\nMissing Values:")
missing_values = df.isnull().sum()
print(missing_values)

print("\nTotal Missing Values:", df.isnull().sum().sum())

print("\nDuplicate Rows:")
duplicate_count = df.duplicated().sum()
print("Number of duplicate rows:", duplicate_count)

print("\nBasic Statistical Summary:")
display(df.describe(include="all").T)

print("\n" + "=" * 80)
print("2. DATA QUALITY CHECK")
print("=" * 80)

print("\nMissing Value Check:")
if df.isnull().sum().sum() == 0:
    print("✓ No missing values found.")
else:
    print("⚠ Missing values detected.")

print("\nDuplicate Check:")
if duplicate_count == 0:
    print("✓ No duplicate records found.")
else:
    print(f"⚠ {duplicate_count} duplicate records found.")

print("\nData Types:")
for column in df.columns:
    print(f"{column}: {df[column].dtype}")

stats_variables = [
    "Weekly_Study_Hours",
    "Average_Sleep_Hours",
    "Daily_Screen_Time_Hours",
    "Stress_Score",
    "Academic_Readiness_Score"
]

print("\n" + "=" * 80)
print("3. DESCRIPTIVE STATISTICS")
print("=" * 80)

print("""
Formula:
Mean = Σx / n

Median = Middle value after sorting the data

Mode = Most frequently occurring value

Range = Maximum - Minimum

Variance = Σ(x - mean)² / (n - 1)

Standard Deviation = √Variance

IQR = Q3 - Q1
""")

statistics = []

for column in stats_variables:

    mean_value = df[column].mean()
    median_value = df[column].median()

    mode_values = df[column].mode()

    if len(mode_values) == 0:
        mode_value = np.nan
    else:
        mode_value = ", ".join(
            [str(round(x, 3)) for x in mode_values.tolist()]
        )

    min_value = df[column].min()
    max_value = df[column].max()

    range_value = max_value - min_value

    variance_value = df[column].var()
    std_value = df[column].std()

    q1 = df[column].quantile(0.25)
    q3 = df[column].quantile(0.75)

    iqr = q3 - q1

    statistics.append([
        column,
        mean_value,
        median_value,
        mode_value,
        min_value,
        max_value,
        range_value,
        variance_value,
        std_value,
        q1,
        q3,
        iqr
    ])

stats_df = pd.DataFrame(
    statistics,
    columns=[
        "Variable",
        "Mean",
        "Median",
        "Mode",
        "Minimum",
        "Maximum",
        "Range",
        "Variance",
        "Standard Deviation",
        "Q1",
        "Q3",
        "IQR"
    ]
)

display(stats_df.round(3))

print("\n" + "=" * 80)
print("4. STATISTICAL INTERPRETATIONS")
print("=" * 80)

for column in stats_variables:

    mean_value = df[column].mean()
    median_value = df[column].median()
    std_value = df[column].std()
    q1 = df[column].quantile(0.25)
    q3 = df[column].quantile(0.75)
    iqr = q3 - q1

    print(f"\n{column}")
    print("-" * 60)
    print(f"Mean               : {mean_value:.3f}")
    print(f"Median             : {median_value:.3f}")
    print(f"Mode               : {df[column].mode().tolist()}")
    print(f"Range              : {df[column].max() - df[column].min():.3f}")
    print(f"Variance           : {df[column].var():.3f}")
    print(f"Standard Deviation : {std_value:.3f}")
    print(f"Q1                 : {q1:.3f}")
    print(f"Q3                 : {q3:.3f}")
    print(f"IQR                : {iqr:.3f}")

    if mean_value > median_value:
        print("Interpretation: Mean is greater than median, suggesting some higher values may be influencing the average.")
    elif mean_value < median_value:
        print("Interpretation: Mean is lower than median, suggesting some lower values may be influencing the average.")
    else:
        print("Interpretation: Mean and median are equal, suggesting a relatively balanced distribution.")


print("\n" + "=" * 80)
print("5. VARIABLE WITH GREATEST VARIABILITY")
print("=" * 80)

std_values = df[stats_variables].std()

greatest_variability_variable = std_values.idxmax()
greatest_variability_std = std_values.max()

print("\nStandard Deviations:")
display(std_values.to_frame("Standard Deviation").round(3))

print(
    f"\nVariable with the greatest variability based on standard deviation: "
    f"{greatest_variability_variable}"
)

print(f"Standard deviation = {greatest_variability_std:.3f}")

print(
    "\nInterpretation: A higher standard deviation means the observations "
    "are more spread out around the mean."
)

outlier_variables = [
    "Weekly_Study_Hours",
    "Daily_Screen_Time_Hours",
    "Commute_Time_Minutes",
    "Monthly_Discretionary_Spending"
]

print("\n" + "=" * 80)
print("6. IQR OUTLIER DETECTION")
print("=" * 80)

print("""
IQR Method:

IQR = Q3 - Q1

Lower Bound = Q1 - 1.5 × IQR

Upper Bound = Q3 + 1.5 × IQR

Any value below the lower bound or above the upper bound
is considered an outlier.
""")

outlier_results = {}

for column in outlier_variables:

    q1 = df[column].quantile(0.25)
    q3 = df[column].quantile(0.75)
    iqr = q3 - q1

    lower_bound = q1 - 1.5 * iqr
    upper_bound = q3 + 1.5 * iqr

    outlier_mask = (
        (df[column] < lower_bound) |
        (df[column] > upper_bound)
    )

    outliers = df.loc[outlier_mask, column]

    outlier_results[column] = {
        "Q1": q1,
        "Q3": q3,
        "IQR": iqr,
        "Lower Bound": lower_bound,
        "Upper Bound": upper_bound,
        "Outliers": outliers.tolist(),
        "Count": len(outliers)
    }

    print(f"\n{column}")
    print("-" * 60)
    print(f"Q1            : {q1:.3f}")
    print(f"Q3            : {q3:.3f}")
    print(f"IQR           : {iqr:.3f}")
    print(f"Lower Bound   : {lower_bound:.3f}")
    print(f"Upper Bound   : {upper_bound:.3f}")
    print(f"Outlier Count : {len(outliers)}")
    print(f"Outliers      : {outliers.tolist()}")

    if len(outliers) == 0:
        print("Interpretation: No IQR outliers were detected.")
    else:
        print("Interpretation: Values outside the IQR bounds are potential outliers.")

print("\n" + "=" * 80)
print("7. MEAN AND MEDIAN BEFORE/AFTER OUTLIER REMOVAL")
print("=" * 80)

comparison_variable = "Weekly_Study_Hours"

q1 = df[comparison_variable].quantile(0.25)
q3 = df[comparison_variable].quantile(0.75)
iqr = q3 - q1

lower_bound = q1 - 1.5 * iqr
upper_bound = q3 + 1.5 * iqr

cleaned_variable = df[
    (df[comparison_variable] >= lower_bound) &
    (df[comparison_variable] <= upper_bound)
][comparison_variable]

before_mean = df[comparison_variable].mean()
before_median = df[comparison_variable].median()

after_mean = cleaned_variable.mean()
after_median = cleaned_variable.median()

comparison_df = pd.DataFrame({
    "Measure": ["Mean", "Median"],
    "Before Outlier Removal": [before_mean, before_median],
    "After Outlier Removal": [after_mean, after_median],
    "Change": [
        after_mean - before_mean,
        after_median - before_median
    ]
})

display(comparison_df.round(3))

print(
    f"\nInterpretation: For {comparison_variable}, removing IQR outliers "
    "changes the mean/median because extreme observations can influence "
    "the center of the distribution."
)

print("\n" + "=" * 80)
print("8. PROBABILITY ANALYSIS")
print("=" * 80)

print("""
Events:

A = Student has Part_Time_Job = "Yes"

B = Student has Stress_Score >= 7

C = Student has Scholarship = "Yes"

D = Student exercises at least 3 days per week
""")

print("\nPart_Time_Job unique values:")
print(df["Part_Time_Job"].unique())

print("\nScholarship unique values:")
print(df["Scholarship"].unique())

exercise_column_candidates = [
    "Exercise_Days_Per_Week",
    "Exercise_Days",
    "Weekly_Exercise_Days",
    "Exercise_Frequency"
]

exercise_column = None

for col in exercise_column_candidates:
    if col in df.columns:
        exercise_column = col
        break

if exercise_column is None:
    raise ValueError(
        "Exercise column not found. Please check the dataset column name."
    )

print(f"\nExercise column used: {exercise_column}")

A = df["Part_Time_Job"].astype(str).str.strip().str.lower() == "yes"

B = df["Stress_Score"] >= 7

C = df["Scholarship"].astype(str).str.strip().str.lower() == "yes"

D = df[exercise_column] >= 3

n = len(df)


P_A = A.mean()
P_B = B.mean()
P_C = C.mean()
P_D = D.mean()

print("\n" + "=" * 80)
print("9. BASIC EVENT PROBABILITIES")
print("=" * 80)

print(f"P(A) = P(Part_Time_Job = Yes)")
print(f"     = {A.sum()} / {n}")
print(f"     = {P_A:.4f} ({P_A*100:.2f}%)")

print(f"\nP(B) = P(Stress_Score >= 7)")
print(f"     = {B.sum()} / {n}")
print(f"     = {P_B:.4f} ({P_B*100:.2f}%)")

print(f"\nP(C) = P(Scholarship = Yes)")
print(f"     = {C.sum()} / {n}")
print(f"     = {P_C:.4f} ({P_C*100:.2f}%)")

print(f"\nP(D) = P(Exercise >= 3 days/week)")
print(f"     = {D.sum()} / {n}")
print(f"     = {P_D:.4f} ({P_D*100:.2f}%)")


A_or_B = A | B
A_and_B = A & B

P_A_or_B = A_or_B.mean()
P_A_and_B = A_and_B.mean()

print("\n" + "=" * 80)
print("10. UNION AND INTERSECTION")
print("=" * 80)

print("""
Formula:

P(A or B) = P(A ∪ B)
           = P(A) + P(B) - P(A ∩ B)

P(A and B) = P(A ∩ B)
""")

print(f"P(A and B)")
print(f"= {A_and_B.sum()} / {n}")
print(f"= {P_A_and_B:.4f} ({P_A_and_B*100:.2f}%)")

print(f"\nP(A or B)")
print(f"= P(A) + P(B) - P(A and B)")
print(f"= {P_A:.4f} + {P_B:.4f} - {P_A_and_B:.4f}")
print(f"= {P_A_or_B:.4f} ({P_A_or_B*100:.2f}%)")


P_A_given_B = P_A_and_B / P_B
P_B_given_A = P_A_and_B / P_A

print("\n" + "=" * 80)
print("11. CONDITIONAL PROBABILITIES")
print("=" * 80)

print("""
Formula:

P(A|B) = P(A ∩ B) / P(B)

P(B|A) = P(A ∩ B) / P(A)
""")

print(f"P(A|B)")
print(f"= {P_A_and_B:.4f} / {P_B:.4f}")
print(f"= {P_A_given_B:.4f} ({P_A_given_B*100:.2f}%)")

print(f"\nP(B|A)")
print(f"= {P_A_and_B:.4f} / {P_A:.4f}")
print(f"= {P_B_given_A:.4f} ({P_B_given_A*100:.2f}%)")

# ------------------------------------------------
# 15. MUTUALLY EXCLUSIVE EVENTS
# ------------------------------------------------

year1 = df["Year_of_Study"] == 1
year4 = df["Year_of_Study"] == 4

intersection_years = year1 & year4

print("\n" + "=" * 80)
print("12. MUTUALLY EXCLUSIVE EVENTS")
print("=" * 80)

print("""
Two events are mutually exclusive if:

P(A ∩ B) = 0
""")

print(f"Number of Year 1 students: {year1.sum()}")
print(f"Number of Year 4 students: {year4.sum()}")
print(f"Students simultaneously in Year 1 and Year 4: {intersection_years.sum()}")

if intersection_years.sum() == 0:
    print("\nConclusion: Year_of_Study = 1 and Year_of_Study = 4 are mutually exclusive.")
else:
    print("\nConclusion: They are NOT mutually exclusive.")


product_independence = P_A * P_B

print("\n" + "=" * 80)
print("13. INDEPENDENCE OF A AND B")
print("=" * 80)

print("""
For two events to be independent:

P(A ∩ B) = P(A) × P(B)
""")

print(f"P(A ∩ B)       = {P_A_and_B:.4f}")
print(f"P(A) × P(B)    = {P_A:.4f} × {P_B:.4f}")
print(f"                = {product_independence:.4f}")

difference = abs(P_A_and_B - product_independence)

print(f"\nAbsolute difference = {difference:.4f}")


if np.isclose(P_A_and_B, product_independence, atol=0.01):
    print("\nConclusion: A and B appear approximately independent.")
else:
    print("\nConclusion: A and B do NOT appear independent.")

print(
    "\nInterpretation: Since the observed joint probability differs "
    "substantially from P(A) × P(B), having a part-time job and having "
    "a high stress score appear to be associated in this sample."
)


print("\n" + "=" * 80)
print("14. BAYES' THEOREM")
print("=" * 80)

print("""
Bayes' Theorem:

P(A|B) = [P(B|A) × P(A)] / P(B)

Also:

P(B) = P(B|A)P(A) + P(B|not A)P(not A)
""")

not_A = ~A

P_not_A = not_A.mean()

P_B_given_not_A = (B & not_A).sum() / not_A.sum()

bayes_numerator = P_B_given_A * P_A

bayes_denominator = (
    P_B_given_A * P_A
    +
    P_B_given_not_A * P_not_A
)

P_A_given_B_bayes = bayes_numerator / bayes_denominator

print(f"P(A) = {P_A:.4f}")
print(f"P(B) = {P_B:.4f}")
print(f"P(B|A) = {P_B_given_A:.4f}")
print(f"P(B|not A) = {P_B_given_not_A:.4f}")
print(f"P(not A) = {P_not_A:.4f}")

print("\nBayes calculation:")

print(
    f"\nP(A|B) = "
    f"[{P_B_given_A:.4f} × {P_A:.4f}] / {P_B:.4f}"
)

print(
    f"P(A|B) using Bayes = {P_A_given_B_bayes:.4f} "
    f"({P_A_given_B_bayes*100:.2f}%)"
)

print(
    f"\nDirect P(A|B) = {P_A_given_B:.4f} "
    f"({P_A_given_B*100:.2f}%)"
)

print(
    f"Difference = {abs(P_A_given_B - P_A_given_B_bayes):.10f}"
)

if np.isclose(P_A_given_B, P_A_given_B_bayes):
    print("\n✓ Bayes' theorem agrees with the direct conditional probability.")
else:
    print("\n⚠ Small numerical difference may be present due to rounding.")

print("\n" + "=" * 80)
print("15. NORMAL DISTRIBUTION - ACADEMIC READINESS SCORE")
print("=" * 80)

academic = df["Academic_Readiness_Score"]

academic_mean = academic.mean()
academic_std = academic.std()

highest_index = academic.idxmax()
lowest_index = academic.idxmin()

highest_score = academic.loc[highest_index]
lowest_score = academic.loc[lowest_index]

highest_z = (highest_score - academic_mean) / academic_std
lowest_z = (lowest_score - academic_mean) / academic_std

print("""
Z-score formula:

Z = (X - μ) / σ

Where:
X = individual score
μ = population/sample mean
σ = standard deviation
""")

print(f"Mean Academic Readiness Score = {academic_mean:.3f}")
print(f"Standard Deviation             = {academic_std:.3f}")

print("\nHighest Academic Readiness Score:")
print(f"Score = {highest_score}")
print(f"Z-score = ({highest_score} - {academic_mean:.3f}) / {academic_std:.3f}")
print(f"Z-score = {highest_z:.3f}")

print("\nLowest Academic Readiness Score:")
print(f"Score = {lowest_score}")
print(f"Z-score = ({lowest_score} - {academic_mean:.3f}) / {academic_std:.3f}")
print(f"Z-score = {lowest_z:.3f}")

print("\nZ-score Interpretation:")

if highest_z > 2:
    print(
        f"Highest score interpretation: The highest score is "
        f"{highest_z:.2f} standard deviations ABOVE the mean, "
        "which is unusually high relative to the sample."
    )
else:
    print(
        f"Highest score interpretation: The highest score is "
        f"{highest_z:.2f} standard deviations from the mean."
    )

if lowest_z < -2:
    print(
        f"Lowest score interpretation: The lowest score is "
        f"{abs(lowest_z):.2f} standard deviations BELOW the mean, "
        "which is unusually low relative to the sample."
    )
else:
    print(
        f"Lowest score interpretation: The lowest score is "
        f"{abs(lowest_z):.2f} standard deviations from the mean."
    )

print("\n" + "=" * 80)
print("16. 68-95-99.7 EMPIRICAL RULE")
print("=" * 80)

print("""
For an approximately normal distribution:

Within ±1 standard deviation:
Approximately 68% of observations

Within ±2 standard deviations:
Approximately 95% of observations

Within ±3 standard deviations:
Approximately 99.7% of observations
""")

empirical_rule = pd.DataFrame({
    "Standard Deviations From Mean": [
        "Within ±1 SD",
        "Within ±2 SD",
        "Within ±3 SD"
    ],
    "Expected Percentage": [
        "68%",
        "95%",
        "99.7%"
    ],
    "Lower Limit": [
        academic_mean - academic_std,
        academic_mean - 2 * academic_std,
        academic_mean - 3 * academic_std
    ],
    "Upper Limit": [
        academic_mean + academic_std,
        academic_mean + 2 * academic_std,
        academic_mean + 3 * academic_std
    ]
})

display(empirical_rule.round(3))

print("\nActual sample percentages:")

within_1 = (
    (academic >= academic_mean - academic_std) &
    (academic <= academic_mean + academic_std)
).mean() * 100

within_2 = (
    (academic >= academic_mean - 2 * academic_std) &
    (academic <= academic_mean + 2 * academic_std)
).mean() * 100

within_3 = (
    (academic >= academic_mean - 3 * academic_std) &
    (academic <= academic_mean + 3 * academic_std)
).mean() * 100

print(f"Actual percentage within ±1 SD = {within_1:.2f}%")
print(f"Actual percentage within ±2 SD = {within_2:.2f}%")
print(f"Actual percentage within ±3 SD = {within_3:.2f}%")

print(
    "\nInterpretation: The empirical rule provides expected percentages "
    "for an approximately normal distribution. The actual sample "
    "percentages may differ from these theoretical values."
)

print("\n" + "=" * 80)
print("17. ACADEMIC READINESS SCORE DISTRIBUTION")
print("=" * 80)

plt.figure(figsize=(10, 6))
plt.hist(academic, bins=10, edgecolor="black")
plt.axvline(
    academic_mean,
    linestyle="--",
    linewidth=2,
    label=f"Mean = {academic_mean:.2f}"
)

plt.title("Distribution of Academic Readiness Scores")
plt.xlabel("Academic Readiness Score")
plt.ylabel("Number of Students")
plt.legend()
plt.show()


for column in outlier_variables:

    plt.figure(figsize=(8, 5))
    plt.boxplot(df[column].dropna())
    plt.title(f"Box Plot - {column}")
    plt.ylabel(column)
    plt.show()


print("\n" + "=" * 80)
print("18. FINAL PROBABILITY SUMMARY")
print("=" * 80)

probability_summary = pd.DataFrame({
    "Probability": [
        "P(A)",
        "P(B)",
        "P(C)",
        "P(D)",
        "P(A or B)",
        "P(A and B)",
        "P(A|B)",
        "P(B|A)"
    ],
    "Value": [
        P_A,
        P_B,
        P_C,
        P_D,
        P_A_or_B,
        P_A_and_B,
        P_A_given_B,
        P_B_given_A
    ],
    "Percentage": [
        P_A * 100,
        P_B * 100,
        P_C * 100,
        P_D * 100,
        P_A_or_B * 100,
        P_A_and_B * 100,
        P_A_given_B * 100,
        P_B_given_A * 100
    ]
})

display(probability_summary.round(4))


print("\n" + "=" * 80)
print("19. OUTLIER SUMMARY")
print("=" * 80)

outlier_summary = []

for column in outlier_variables:

    result = outlier_results[column]

    outlier_summary.append([
        column,
        result["Q1"],
        result["Q3"],
        result["IQR"],
        result["Lower Bound"],
        result["Upper Bound"],
        result["Count"],
        result["Outliers"]
    ])

outlier_summary_df = pd.DataFrame(
    outlier_summary,
    columns=[
        "Variable",
        "Q1",
        "Q3",
        "IQR",
        "Lower Bound",
        "Upper Bound",
        "Outlier Count",
        "Outlier Values"
    ]
)

display(outlier_summary_df.round(3))

print("\n" + "=" * 80)
print("20. FIVE MEANINGFUL STATISTICAL OBSERVATIONS")
print("=" * 80)

# Observation 1
print(
    f"""
1. Academic Readiness:
   The average Academic Readiness Score is {academic_mean:.2f},
   with a standard deviation of {academic_std:.2f}. This indicates
   the typical academic readiness level and the spread of scores
   around the mean.
"""
)

# Observation 2
print(
    f"""
2. Greatest Variability:
   {greatest_variability_variable} has the greatest standard deviation
   among the selected descriptive-statistics variables, with a value
   of {greatest_variability_std:.2f}. Therefore, it shows the largest
   spread relative to its mean among these variables.
"""
)

# Observation 3
outlier_counts_text = []

for column in outlier_variables:
    count = outlier_results[column]["Count"]
    outlier_counts_text.append(f"{column}: {count}")

print(
    """
3. Outliers:
   IQR analysis was used to detect unusually high or low observations.
   The detected outlier counts are:
   """
)

for item in outlier_counts_text:
    print("   -", item)

# Observation 4
print(
    f"""
4. Part-Time Jobs and Stress:
   {P_A*100:.2f}% of students have a part-time job, while
   {P_B*100:.2f}% have a Stress Score of at least 7.
   P(A and B) = {P_A_and_B:.4f}, whereas P(A) × P(B)
   = {product_independence:.4f}. The difference suggests that
   the two events do not appear independent in this sample.
"""
)

# Observation 5
print(
    f"""
5. Academic Readiness Extremes:
   The highest Academic Readiness Score has a Z-score of
   {highest_z:.2f}, while the lowest has a Z-score of
   {lowest_z:.2f}. These values indicate that the highest and
   lowest scores are relatively far from the sample mean.
"""
)

print("\n" + "=" * 80)
print("21. OVERALL CONCLUSION")
print("=" * 80)

print(
    f"""
The Student Wellbeing Survey was analyzed using descriptive statistics,
outlier detection, probability, conditional probability, Bayes' theorem,
and normal-distribution concepts.

The mean, median, mode, range, variance, standard deviation, quartiles,
and IQR were calculated for Weekly Study Hours, Average Sleep Hours,
Daily Screen Time, Stress Score, and Academic Readiness Score.

The IQR method was used to identify outliers in Weekly Study Hours,
Daily Screen Time, Commute Time, and Monthly Discretionary Spending.

For probability analysis, the required events were evaluated.
The probability of having a part-time job was {P_A:.4f}, while the
probability of having a Stress Score of at least 7 was {P_B:.4f}.

The joint probability P(A and B) was {P_A_and_B:.4f}, compared with
P(A) × P(B) = {product_independence:.4f}. Therefore, A and B do not
appear independent in this sample.

Bayes' theorem produced P(A|B) = {P_A_given_B_bayes:.4f}, which agrees
with the direct conditional probability calculation of
{P_A_given_B:.4f}.

For Academic Readiness Score, the highest score had a Z-score of
{highest_z:.3f}, and the lowest score had a Z-score of {lowest_z:.3f}.
The empirical rule indicates that approximately 68%, 95%, and 99.7%
of observations are expected to fall within 1, 2, and 3 standard
deviations of the mean respectively.
"""
)

print("\n" + "=" * 80)
print("ANALYSIS COMPLETED SUCCESSFULLY")
print("=" * 80)
```
